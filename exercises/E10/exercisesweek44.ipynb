## Exercise 2:

**a)** Linear and logistic regression methods

1. What is the main difference between ordinary least squares and Ridge regression?
A: OLS fits the training data in a way that only minimizes the MSE or L1 loss. Ridge does the same but has a regularization parameter lambda which penalizes the training when it gains too much variance from the training data.

2. Which kind of data set would you use logistic regression for?
A: For classification tasks.

3. In linear regression you assume that your output is described by a continuous non-stochastic function $f(x)$. Which is the equivalent function in logistic regression?
A: Here we model the conditional probability modeled via the sigmoid. It states the probability of a class given an observation $x$.

4. Can you find an analytic solution to a logistic regression type of problem?
A: No closed form solution exists.

5. What kind of cost function would you use in logistic regression?
A: Corss-entropy loss. 


**b)** Deep learning

1. What is an activation function and discuss the use of an activation function? Explain three different types of activation functions?
A: ReLU, Sigmoid, LeakyReLU are three types of activation functions. These provide the non-linearity needed to learn complex non-linear dependancie in the data.

2. Describe the architecture of a typical feed forward Neural Network (NN).
A: Input layer $x$ -> Fully connected layer with activation functions at each node -> output layer (linear regression, logistic regression) tailored to the specific task at hand. 

3. You are using a deep neural network for a prediction task. After training your model, you notice that it is strongly overfitting the training set and that the performance on the test isn’t good. What can you do to reduce overfitting?
A: I can use more training examples if available, I can generate new training examples using augmentation techniques, I can add more normalization like dropout, or I can reduce the number of layers to reduce the models capacity so it does not memorize the training data.

4. How would you know if your model is suffering from the problem of exploding gradients?
A: Eventually the model training loop would show a very large loss and eventuallt NaN.

5. Can you name and explain a few hyperparameters used for training a neural network?
A: Learning rate is the update step size, dropout is the probability of deactiating a node, batch size is the number of training examples used per epoch.

6. Describe the architecture of a typical Convolutional Neural Network (CNN)
A: the convolution layer extracts learned features, the pooling layer reduces the dimensionality, this is repeated multiple times until the embedding is then fed into a fully connected layer and used for the final task ie. classification with logistic regression for the output layer. 

7. What is the vanishing gradient problem in Neural Networks and how to fix it?
A: The vanishing gradient effect is where gradients from earlier in the network shrink by the time the output reaches the loss function. Leading to the update function taking small slow steps in the right direction. This can be mitigated by using an activation function like ReLU since it does not have the same exponential decay behaviour.

8. When it comes to training an artificial neural network, what could the reason be for why the cost/loss doesn't decrease in a few epochs?
A: The weights are exploring the problem space.

9. How does L1/L2 regularization affect a neural network?
A: Regularization keeps the model from capturing too mch of the variance specific to the training set. This is important for the model to generalize well to unseen data.

10. What is(are) the advantage(s) of deep learning over traditional methods like linear regression or logistic regression?
A: A sort of one-size-fits-all approach allowing us to minimize development time by using the same or similar 'tool' for many different applications. Also the ability to learn high dimensional non-linearities is very powerful.


**c)** Optimization part

1. Which is the basic mathematical root-finding method behind essentially all gradient descent approaches(stochastic and non-stochastic)?
A: minimization of a loss function using Newtons method.

2. And why don't we use it? Or stated differently, why do we introduce the learning rate as a parameter?
A: Because computing the hessian matrix can be very expensive.

3. What might happen if you set the momentum hyperparameter too close to 1 (e.g., 0.9999) when using an optimizer for the learning rate?
A: Convergence could be very slow, and overshoot the true minimum.

4. Why should we use stochastic gradient descent instead of plain gradient descent?
A: We only need to include a mini-batch of $x$ per epoch. This makes the computations much more efficient.

5. Which parameters would you need to tune when use a stochastic gradient descent approach?
A: Batch size, learning rate, momentum if applicable. 


**d)** Analysis of results

1. How do you assess overfitting and underfitting?
A: Overfitting can be identified by using a validation set or a holdout set. This is not used for training so if the model performs better on the training set compared to the validation set, we have likely overfit to the training data.

2. Why do we divide the data in test and train and/or eventually validation sets?
A: To evaluate the generalizability of our model during or after training. The validation set can be used to tun hyperparameters, while the test set validates the final performance of the model. We can not use the test set to pick the 'best' model.

3. Why would you use resampling methods in the data analysis? Mention some widely popular resampling methods.

4. Why might a model that does not overfit the data (maybe because there is a lot of data) perform worse when we add regularization?
A: Adding regularization to a well-performing model will just increase its bias, in turn making it perform worse.
