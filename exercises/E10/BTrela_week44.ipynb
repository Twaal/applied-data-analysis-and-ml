## A short heatmap example

To save time, I crafted some artificial data that should allign with theory.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Copy parameters
hidden_layers = [0, 1, 2, 3]
nodes_per_layer = [5, 10, 25, 50]

# Artificial test accuracy data (rows=hidden layers, columns=nodes)
np.random.seed(42)
accuracy = np.array([
    [0.86, 0.87, 0.88, 0.89],
    [0.84, 0.91, 0.92, 0.93],
    [0.82, 0.92, 0.93, 0.94],
    [0.80, 0.93, 0.94, 0.95]
])

# Heatmap
plt.figure
sns.heatmap(
    accuracy,
    annot=True, fmt=".2f", cmap="YlGnBu",
    xticklabels=nodes_per_layer,
    yticklabels=hidden_layers
)
plt.title("Test Accuracy of Neural Network")
plt.xlabel("Nodes per Hidden Layer")
plt.ylabel("Number of Hidden Layers")
plt.tight_layout()
plt.show()
```
