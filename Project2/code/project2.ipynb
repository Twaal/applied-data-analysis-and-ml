Here we train our network and compare the results with those from our OLS

```python
# Baseline (scaled): OLS on Runge with polynomial features built on scaled x, predicting scaled y
import numpy as np
import matplotlib.pyplot as plt
import p2_nn_module as p2

# Polynomial degree (keep consistent with your Project 1 best)
degree = 10

# Build polynomial design matrices on the scaled train/test splits
X_train_poly = p2.polynomial_features(X_train_s.ravel(), degree)
X_test_poly  = p2.polynomial_features(X_test_s.ravel(), degree)

# Fit OLS parameters on scaled targets
beta = p2.OLS_parameters(X_train_poly, y_train_s)
print(f"beta shape = {beta.shape}")

# Predictions in scaled target space
y_train_pred_s = X_train_poly @ beta
y_test_pred_s  = X_test_poly  @ beta

# MSE in scaled space
mse_train_s = p2.MSE(y_train_s, y_train_pred_s)
mse_test_s  = p2.MSE(y_test_s,  y_test_pred_s)
print({"mse_train_scaled": mse_train_s, "mse_test_scaled": mse_test_s})

# Visualization in scaled space
# Prepare a dense grid in original x, then scale x and true y to the same space used for training
xx = np.linspace(-1.0, 1.0, 500).reshape(-1, 1)
xx_s = x_scaler.transform(xx)
Xx_poly = p2.polynomial_features(xx_s.ravel(), degree)
yy_true = p2.runge(xx.ravel()).reshape(-1, 1)
yy_true_s = y_scaler.transform(yy_true).ravel()
yy_ols_s  = (Xx_poly @ beta).ravel()

plt.figure(figsize=(7,4))
plt.plot(xx_s.ravel(), yy_true_s, 'k-', label='Runge (true, scaled)')
plt.plot(xx_s.ravel(), yy_ols_s,  'r--', label=f'OLS poly deg {degree} (scaled)')
plt.scatter(X_test_s.ravel(), y_test_s.ravel(), s=12, alpha=0.5, label='Test data (scaled)')
plt.xlabel('x (scaled)')
plt.ylabel('y (scaled)')
plt.title('Runge function: OLS polynomial baseline (scaled space)')
plt.legend()
plt.tight_layout()
plt.show()
```

### Discussion: activation functions, depth, and overfitting
- Sigmoid tends to train more slowly and can saturate for inputs far from zero, often requiring more epochs or careful scaling; on Runge, ReLU/LeakyReLU usually reach lower MSE more quickly.
- ReLU is a strong default; it learns sharp changes efficiently but can suffer from dead neurons if learning rate is high; LeakyReLU mitigates dead neurons with a small negative slope and often matches or slightly improves over ReLU.
- Increasing depth and width reduces bias (lower train MSE) but increases variance; watch the generalization gap (Test − Train MSE) heatmaps: larger gaps at higher depth/width signal overfitting.
- Preferred regime on this 1D task is typically shallow (1–2 layers) with moderate width; very deep/wide networks can overfit without clear gains in Test MSE.
- If you observe overfitting: try early stopping (use the learning curves), reduce width/depth, add L2 (weight decay), or lower the learning rate; for stability across runs, keep the StandardScaler and random seed fixed.

## How we measure accuracy here

We report overall classification accuracy computed from the confusion matrix. Let $\mathrm{CM}$ be the (unnormalized) confusion matrix, where $\mathrm{CM}_{ij}$ counts how many samples of true class $i$ were predicted as class $j$. The diagonal entries $\mathrm{CM}_{ii}$ are correct predictions.

- Overall accuracy is the fraction of correct predictions among all samples:

$$
\text{Accuracy} 
= \frac{\sum_{i=1}^{K} \mathrm{CM}_{ii}}{\sum_{i=1}^{K}\sum_{j=1}^{K} \mathrm{CM}_{ij}}
= \frac{\mathrm{trace}(\mathrm{CM})}{\sum \mathrm{CM}}.
$$

- In code, this equals `sklearn.metrics.accuracy_score(y_true, y_pred)` and also matches `np.trace(CM) / CM.sum()` when `CM` is the raw count matrix.

Notes on normalization and display:
- We often display a normalized confusion matrix for readability (e.g., each row sums to 1). Normalization affects only the displayed cell values, not how accuracy is computed.
- The accuracy number is always computed from the raw counts before normalization.
